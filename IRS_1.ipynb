```python
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.metrics.pairwise import cosine_similarity
doc1="machine learning is a branch of artificial intelligence that focues on building applications that learn erom data."
doc2="Artificial intilligence and machine learning are widely used to make smart applications based on data."
documents=[doc1,doc2]
vectorizer=TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(documents)
similarity_score=cosine_similarity(tfidf_matrix[0:1],tfidf_matrix[1:2])[0][0]
print("cosine _similarity between docment 1 and document 2:")
print(f"{similarity_score:.4f}")
#
```
